1. [20 pts] Explain the difference between a robot architecture and a single robot control
program. Why is architecture more important as a robot becomes more autonomous?

A robot control program tells a robot to do something like follow a line or track an object. Robot architecture is a larger organizational framework guiding perception, decision making, and action. Architecture becomes more important as tasks become more complex becuase the robot will need to reason about its task. Architecture keeps complicated behaviors organized and reliable.

2. [20 pts] Consider a mobile robot that follows a target object using a camera and avoids
obstacles using a distance sensor.
Identify the major subsystems and describe how information flows from sensing to action.
Your answer should include perception, control, actuation, and power/communication
concerns.

The major subsystems are perception, control, actuation, power/communication. 
- Perception: camera and sensors detects and locates objects
- Control: uses perception data to decide how robot should act based on perception data
- Actuation: control systems send signals to actuation to make the robot act accordingly.
- power/comms: battery supplies power to all systems, communications allows systems to communicate via some protocol like UART or CAN

3. [20 pts] Compare reactive and deliberative control for the target-following robot.
Give one situation where reactive control is better and one situation where deliberative
control is better.

Reactive control allows the robot to react to something it senses, like a human stepping in its way, so it can act accordingly. This is good for things like self driving cars, where the car must make a decision quickly based on an obstacle

Deliberative control plans actions before executing them. This is better suited for something like a robot arm on an assembly line, where the motions are predetermined and repeatalble and just need to be executed


4. [20 pts] Suppose the camera detects the target object, but the robot occasionally turns the
wrong direction.
Give three possible causes from different architecture layers, for example sensor,
perception, controller, actuator, or communication. For each cause, suggest one diagnostic
test.

Perception: the sensor may be recieving a lot of noise and the algorithm that percieves may make the wrong decision of what target its tracking. Best diagnostic is to actually view what the sensor is seeing and tracking over time.

Control: Steering logic may contain a sign error where movement is inverted. Best diagnostic is to feed known data/actions into the system and see how it reacts. 

Actuator/comms: The motors or actuators may be recieved a bad signal from another part of the system, or the wrong motors are recieving correct signals. Diagnostic: command each motor individually with known signals to make sure they move accordingly.
